### 1-2. 문자열을 숫자로 변환하기
---

**요구사항**

① 빈 리스트 `houses`를 생성합니다.

② 반복문을 사용하여 각 행을 순회합니다.

③ 각 행의 값을 아래 자료형으로 변환하여 `houses`에 저장합니다.

| 열 이름 | 변환할 자료형 |
| --- | --- |
| `ID` | `int` |
| `size` | `int` |
| `year` | `int` |
| `price` | `float` |

④ 필수 값이 누락되었거나 숫자로 변환할 수 없는 행은 `houses`에 저장하지 않고, 행 위치와 제외 사유를 출력합니다.

⑤ 한 행에서 변환 오류가 발생해도 반복문을 중단하지 않고 다음 행을 계속 처리한 뒤, 정상 저장 행 수와 제외 행 수를 출력합니다.

**✅ 확인 포인트**

- `houses`는 빈 리스트로 생성되어야 합니다.
- 각 행은 반복문으로 처리되어야 합니다.
- `ID`, `size`, `year`는 `int`로 변환되어야 합니다.
- `price`는 `float`로 변환되어야 합니다.
- 누락·변환 오류 행의 위치와 제외 사유, 정상 저장 행 수와 제외 행 수가 출력되어야 합니다.

####문제 2  
---
실행 결과 :

````python
# ① 빈 리스트 houses를 생성합니다.
houses = []
e_tmp=[]    # 제외한 행의 위치를 저장

# 파일을 정상적으로 읽었을 때만 자료형 변환을 진행
if df_house is not None:
    # ② 반복문을 사용하여 각 행을 순회합니다.
    for i, row in df_house.iterrows():
        # ⑤ 한 행에서 변환 오류가 발생해도 반복문을 중단하지 않고 다음 행을 계속 처리
        try:
            # ④ 필수 값이 누락되었으면 해당 행을 제외합니다.
            if row[['ID', 'size', 'year', 'price']].isna().any():
                raise ValueError('필수 값이 누락되었습니다.')

            # ③ 각 행의 값을 지정된 자료형으로 변환하여 houses에 저장합니다.
            # 문제 3에서 열 이름으로 값을 검사하므로 주택 한 채는 딕셔너리로 저장합니다.
            houses.append({
                'ID': int(row['ID']),
                'size': int(row['size']),
                'year': int(row['year']),
                'price': float(row['price'])
            })
        except (ValueError, TypeError, KeyError, OverflowError) as error_message:
            # ④ 저장하지 않은 행의 위치와 제외 사유를 출력합니다.
            e_tmp.append(i)
            print(f'제외된 행 위치(Index): {i}, 제외 사유: {error_message}')
else:
    print('정상 CSV 파일을 먼저 읽어 주세요.')

# ⑤ 정상 저장 행 수와 제외 행 수를 출력합니다.
print('## 저장 및 제외된 행 출력')
print(f'\t정상 저장 행 수 : {len(houses)}')
print(f'\t제외 행 수 : {len(e_tmp)}')
````

- `size`가 30 이상
- `price`가 4 이하

⑥ 선택한 주택의 ID를 하나씩 출력합니다.

⑦ 선택한 주택의 전체 건수를 출력합니다.

**✅ 확인 포인트**

- `size`가 0 이하이거나 `price`가 음수인 주택은 `selected`에 포함되지 않아야 합니다.
- `size`가 30 이상이고 `price`가 4 이하인 주택만 `selected`에 저장되어야 합니다.
- 선택된 주택의 ID가 각각 출력되어야 합니다.
- 선택된 주택의 전체 건수가 출력되어야 합니다.
- 힌트

    - 두 조건을 모두 만족하는지 검사할 때는 and 사용.

####문제 3  
---
실행 결과 :

````python
# ① 원본 houses를 보존하고 복사본에 테스트 데이터를 추가합니다.
houses_test = [house.copy() for house in houses]
houses_test.extend([
    {'ID': 101, 'year': 5, 'price': 3},                 # size 키 없음 테스트
    {'ID': 102, 'size': 35, 'year': 5, 'price': None},  # 값 없음 테스트
    {'ID': 103, 'size': '오류', 'year': 5, 'price': 3}, # 숫자 변환 실패 테스트
    {'ID': 104, 'size': 35, 'year': 5, 'price': 3},     # 정상 테스트 데이터
])
# display(houses_test)

# 빈 리스트를 준비
selected = []

print('## 데이터 검사 중')

# ② 테스트 주택을 for문으로 하나씩 검사합니다.
for house in houses_test:
    # ③ 필요한 키가 없으면 사유를 출력하고 현재 항목을 건너뜁니다.
    if 'size' not in house or 'price' not in house:
        print(f"\tID ({house.get('ID')}) 제외\t(제외 사유: size 또는 price 키가 없습니다.)")
        continue

    # ③ 값이 비어 있으면 조건 비교 전에 제외합니다.
    if pd.isna(house['size']) or pd.isna(house['price']):
        print(f"\tID ({house.get('ID')}) 제외\t(제외 사유: 값이 비어 있습니다.)")
        continue

    # ③ 숫자 형태 문자열은 숫자로 바꾸고, 변환 실패는 건너뜁니다.
    try:
        size = float(house['size'])
        price = float(house['price'])
    except (ValueError, TypeError):
        print(f"\tID ({house.get('ID')}) 제외\t(제외 사유: 숫자로 변환할 수 없습니다.)")
        continue

    # 요구사항 3-④, 3-⑤: size가 0 이하 또는 price가 음수이면 제외합니다.
    if size <= 0 or price < 0:
        print(f"\tID ({house.get('ID')}) 제외\t(제외 사유: 유효하지 않은 값입니다.)")
        continue

    # ④ 두 조건을 모두 만족해야 하므로 and를 사용해 저장합니다.
    if size >= 30 and price <= 4:
        selected.append(house)

print('')
print('## size가 30 이상 & price가 4 이하 조건을 만족하는 ID 출력')
# ⑥ 선택된 주택의 ID를 하나씩 출력합니다.
for house in selected:
    print('\t선택된 주택 ID:', house['ID'])

# ⑦ 선택된 주택의 전체 건수를 출력합니다.
print('\t선택된 주택의 전체 건수:', len(selected))
````

실행 결과 :


````python
# ① 크기당 가격을 계산하는 함수를 작성합니다.
def unit_price(price, size):
    # ② 숫자 또는 숫자 형태의 문자열을 입력받도록 작성합니다.


    # ⑦ 잘못된 입력이 있어도 프로그램이 중단되지 않도록 처리합니다.
    try:

        price = float(price)
        size = float(size)

        # ④ 크기가 0 이하이거나 가격이 음수이면 각각 원인을 알 수 있는 안내 문구를 출력하고 None을 반환합니다.
        if size <= 0 or price < 0:
            if size <= 0:
                print('Error:\t크기 오류: 크기는 0보다 커야 합니다.')
            else:
                print('Error:\t가격 오류: 가격은 음수가 될 수 없습니다.')
            return None
        else:
            # ③ 정상 입력이면 가격 ÷ 크기를 반환합니다.
            return price / size

        # ⑤ 숫자로 변환할 수 없는 문자열을 입력하여 ValueError가 발생하면,
        #   “값 오류: 숫자로 변환할 수 없는 문자열입니다.”라는 안내 문구를 출력하고 None을 반환합니다.
    except ValueError:
        print('Error:\t값 오류: 숫자로 변환할 수 없는 문자열입니다.')
        return None

        # ⑥ None 등 숫자로 변환할 수 없는 자료형을 입력하여 TypeError가 발생하면,
        #   “자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.”라는 안내 문구를 출력하고 None 을 반환합니다.
    except TypeError:
        print('Error:\t자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.')
        return None

# ⑧ 작성한 함수를 전체 houses에 적용하여 주택 ID와 크기당 가격을 출력합니다.
print('## 주택 ID와 크기당 가격을 출력')
for house in houses:
    print(f'\t주택 ID: {house["ID"]}, 크기당 가격: {unit_price(house["price"], house["size"])}')

print('')
# ⑨ 아래 호출을 순서대로 실행하여 반환값과 안내 문구를 확인합니다.
#   함수 호출 반환값과 안내 문구를 출력.
print('## 주택 ID와 크기당 가격을 출력')
print('\t1. unit_price(4, 40)')
print('\t',unit_price(4, 40))

print('\t2. unit_price("4", "40")')
print('\t',unit_price("4", "40"))

print('\t3. unit_price(4, 0)')
print('\t',unit_price(4, 0))

print('\t4. unit_price(-4, 40)')
print('\t',unit_price(-4, 40))

print('\t5. unit_price("오류", 40)')
print('\t',unit_price("오류", 40))

print('\t6. unit_price(None, 40)')
print('\t',unit_price(None, 40))

print('\t7. unit_price(6, 30)')
print('\t',unit_price(6, 30))
````
